```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

plt.style.use('seaborn-v0_8-darkgrid')
%matplotlib inline
```

```python
#load international matches data
matches = pd.read_csv('../data/raw/results.csv')

#display first few rows of the dataframe
matches.head()
```

```python
print(f"Total matches: {len(matches)}")
print(f"Date range: {matches['date'].min()} to {matches['date'].max()}")
matches.info()
```

```python
goalscorers = pd.read_csv('../data/raw/goalscorers.csv')
goalscorers.head()
```

```python
ronaldo_goals = goalscorers[goalscorers['scorer'].str.contains('Cristiano Ronaldo', case=False, na=False)]
print(f"Ronaldo goals found: {len(ronaldo_goals)}")
ronaldo_goals.head()
```

```python
matches_por = matches[(matches['home_team'] == 'Portugal') | (matches['away_team'] == 'Portugal')]
#check number of matches involving Portugal
print(f"Total matches involving Portugal: {len(matches_por)}")
```

```python
matches_por_w = matches[
    (matches['home_team'] == 'Portugal') & (matches['home_score'] > matches['away_score']) | 
    (matches['away_team'] == 'Portugal') & (matches['away_score'] > matches['home_score'])]
#winrate calculation
winrate_por = len(matches_por_w) / len(matches_por)
print(f"Portugal winrate: {winrate_por}")
```

```python
matches['tournament'].unique()
```

```python
matches_wc = matches[matches['tournament'] == 'FIFA World Cup']
```

```python
matches_por_wc = matches_wc[(matches_wc['home_team'] == 'Portugal') | (matches_wc['away_team'] == 'Portugal')]
print(f"Total World Cup matches involving Portugal: {len(matches_por_wc)}")
#find winrate in World Cup matches
matches_por_wc_w = matches_wc[
    (matches_wc['home_team'] == 'Portugal') & (matches_wc['home_score'] > matches_wc['away_score']) | 
    (matches_wc['away_team'] == 'Portugal') & (matches_wc['away_score'] > matches_wc['home_score'])]
winrate_por_wc = len(matches_por_wc_w) / len(matches_por_wc)
print(f"Portugal World Cup winrate: {winrate_por_wc}")
matches_por_wc = matches_por_wc.sort_values('date', ascending=False)
matches_por_wc.iloc[:10]
ronaldo_goals_wc = ronaldo_goals[ronaldo_goals['date'].isin(matches_por_wc['date'])]
print(f"Ronaldo World Cup goals found: {len(ronaldo_goals_wc)}")
# Ronaldo-era Portugal WC matches (2006 onwards)
matches_wc_ronaldo_era = matches_wc[
    ((matches_wc['home_team'] == 'Portugal') | (matches_wc['away_team'] == 'Portugal')) &
    (matches_wc['date'] >= '2006-01-01')
]
print(f"Portugal WC matches in Ronaldo era: {len(matches_wc_ronaldo_era)}")
#ronaldo wc scoring rate
ronaldo_goals_wc_era = ronaldo_goals_wc[ronaldo_goals_wc['date'].isin(matches_wc_ronaldo_era['date'])]
ronaldo_wc_scoring_rate = len(ronaldo_goals_wc_era) / len(matches_wc_ronaldo_era)
print(f"Ronaldo World Cup scoring rate (goals per match): {ronaldo_wc_scoring_rate}")
```

```python
#find colombia's strength (winrate, goals scored, etc.)
matches_col = matches[
    (matches['home_team'] == 'Colombia') | (matches['away_team'] == 'Colombia')]
matches_col_w = matches[(matches['home_team'] == 'Colombia') & (matches['home_score'] > matches['away_score']) | 
                        (matches['away_team'] == 'Colombia') & (matches['away_score'] > matches['home_score'])]
winrate_col = len(matches_col_w) / len(matches_col)
print(f"Colombia winrate: {winrate_col}")
goals_col = 0
for _, match in matches_col.iterrows():
    if match['home_team'] == 'Colombia':
        goals_col += match['home_score']
    else:
        goals_col += match['away_score']
avg_goals_col = goals_col / len(matches_col)
print(f"Colombia average goals per match: {avg_goals_col}")
```
